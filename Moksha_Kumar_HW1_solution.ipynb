# Question 3 [10+10+4+6]

Speed guns work on the principle of Doppler effect. If the original frequency is $f_0$ and the shift in the received frequency is $\Delta f$, then the speed of the reflector (car in this case) is $v = \dfrac{\Delta f}{f_0} \times \dfrac{c}{2}$, where, $c=3\times10^{10}\,\rm{cm\ s}^{-1}$ is the speed of light. In the file `speedgundata.csv` you will find the readings of Doppler shift $\dfrac{\Delta f}{f_0}$ measured for $1000$ cars each on Atal Setu and Eastern Expressway.  
1. Write function(s) that can take the filename as an argument, reads the data, converts the raw data to **car speed in $\rm{km\ h}^{-1}$**, stores this in a neat dictionary, and returns the dictionary. Run this function to get the car speeds in $\rm{km\ h}^{-1}$ observed on these two roads. 
2. Import the relevant matplotlib module to plot the histograms for the distribution of measured car speeds on the two roads on a single panel. 
3. Speed limits are $100\,\rm{km\ h}^{-1}$ and $80\,\rm{km\ h}^{-1}$ on Atal Setu and Eastern Expressway. Assume that cars that are within $5\%$ above the speed limit do not get speeding tickets (fines). What fraction of cars will get speeding tickets on each road based on this data? 
4. Make a single-panel figure that clearly demonstrates the fraction of cars on each road that will get speeding tickets. 

**Note:** **Do not** use any external modules except `matplotlib`. You will be showing multiple data on single panel figures. Ensure that both are clearly visible. Use labels to ensure that we understand which data is for which road. Do not forget to label the axes with proper units. Marks will depend on how clear and informative your figures are. 

````python
#Part 1

#The function for getting the speeds from the file provided is: -
def create_data(filename):
    freqdata = {'Atal Setu':[],
                'Eastern Expressway':[]    
    }
    with open(filename, 'r') as fread:
        line = fread.readline() #read one line at a time
        for line in fread:
            freqdata['Atal Setu'].append(line.split(',')[0])
            freqdata['Eastern Expressway'].append(line.split(',')[1][:-1])
            mult_factor = 3*(10**8)*18/10
    freqdata['Atal Setu'] = [float(i)*mult_factor for i in freqdata['Atal Setu']]
    freqdata['Eastern Expressway'] = [float(i)*mult_factor for i in freqdata['Eastern Expressway']]
    freqdata['Atal Setu'] = [round(i,2) for i in freqdata['Atal Setu']]
    freqdata['Eastern Expressway'] = [round(i,2) for i in freqdata['Eastern Expressway']]
    return(freqdata)
data = create_data('speedgundata.csv')
````

````python
#Part 2
#Histogram for the datas in two cities.

import matplotlib.pyplot as plt

speed_data = create_data('speedgundata.csv')

plt.figure(figsize=(9, 5))

# Extracting directly using dictionary keys
plt.hist(speed_data['Atal Setu'], bins=30, alpha=0.6, label='Atal Setu', color='navy', edgecolor="k")
plt.hist(speed_data['Eastern Expressway'], bins=30, alpha=0.6, label='Eastern Expressway', color='orange', edgecolor="k")

plt.xlabel('Speed (km/h)')
plt.ylabel('Number of Cars')
plt.title('Speed Distribution')
plt.legend()

plt.show()
````

````python
#Part 3

speed_limit_as = 100 + 0.05*100    #effective speed limit in atal setu
speed_limit_ee = 80 + 0.05*80      #effective speed limit in eastern expressway

fined_cars_as = []    #Speed of Cars in atal setu subject to fine
fined_cars_ee = []    #Speed of Cars in eastern expressway subject to fine

for i in data['Atal Setu']:
    if i > speed_limit_as:
        fined_cars_as.append(i)
    
for j in data['Eastern Expressway']:
    if j > speed_limit_ee:
        fined_cars_ee.append(j)

print("The fraction of cars in Atal Setu gettin' fined is ", len(fined_cars_as)/len(data['Atal Setu']))
print("The fraction of cars in Atal Setu gettin' fined is ", len(fined_cars_ee)/len(data['Eastern Expressway']))
````

````python
#Part 4
# Best thing we can do is create a vertical line seperating fined cars and safe cars.
plt.figure(figsize=(10, 5))
plt.hist(data['Atal Setu'], bins=30, alpha=0.5, label='Atal Setu', color='blue')
plt.hist(data['Eastern Expressway'], bins=30, alpha=0.5, label='Eastern Expressway', color='orange')

# Draw vertical lines for the ticket threshold limits
plt.axvline(speed_limit_as, color='blue', linestyle='--', label='Atal Limit (105 km/h)')
plt.axvline(speed_limit_ee, color='red', linestyle='--', label='Eastern Limit (84 km/h)')

# Basic labels
plt.xlabel('Speed (km/h)')
plt.ylabel('Number of Cars')
plt.title('Car Speed Distribution and Speeding Limits')
plt.legend()
plt.grid(True)

plt.show()
````

Ants finding food is a fascinating dynamics. They sense food using their antennae. As they move towards it, they leave pheromones for other ants to follow. It involves a large number of ants and feddback communications. Here we will simplify considerably and consider only one ant scout and analyse how it finds the food. We will not worry about how it will guide the other ants during its search and travel. 

Here is the simplified model. Consider an ant trying to find food at a distance $d$ on a two-dimensional plane (given by say $\hat{x}$-$\hat{y}$). The angle between the vector connecting the ant and food $\vec{d}$, and $\hat{x}$ is $\mu_\theta$. The ant can sense the rough direction of food. The guessed direction is distributed as normal with a mean $\mu_\theta$ and standard deviation $\sigma_\theta(d)$; as the ant approaches food, the guess improves- $\sigma_\theta=md$, where $m$ is a constant factor. The ant takes steps of unit length. If $d<1$, the ant reaches food in the next step. [Can you guess what will be the problem if you do not assume this?] 

For this problem, you can use any functions from `numpy`. In particular, you may need `random`,`array`, and suitable trigonometric functions to make it efficient. 

1. Write a function that models this motion. Make this code as general as possible. Decide how you will supply the arguments of this function. Write a suitable docstring and enough comments to make it easily understandable. This should, at minimum (but not limited to), return the location coordinates of the ant as it moves and finds the food. For safety, write the function such that it gives up on the ant if it cannot find the food within the first $N$ steps. 
2. Plot the motion of an ant using `food_pos=[10, 20]`, `m=0.1`, ant's starting point `[0,0]`, and `N=1000`. Clearly mark the starting point of the ant and the location of the food. Do not forget to give *all* necessary labels to make the figure understandable.    
3. Write the necessary code/function to run this ant-walking function for $100$ successful ants using the same `food_pos`, ant's starting point, `m`, and `N`. Find the number of steps taken by each successful ant. What is the mean number of steps and standard deviation? 
4. How many unsuccessful ants did you get? 
5. Plot the normalised histogram for the number of steps to find food based on your experiment in `part 3`. 
6. Repeat the same experiment (`part 3`) for `food_pos=[0,5], [5,5], [10,5], [5,10]` keeping everything else the same. Use `scatter` and `errorbar` routines to plot the number of steps as a function of initial $d$ (include the result of your first experiment). Show the mean and standard deviation of the distribution of number of steps taken by successful ants as a point and errorbars, respectively. 
7. Does it show any trends? Does it make sense?

````python
#Part 1

import numpy as np
import matplotlib.pyplot as plt

def ant_walker_function(m=0.1, N=1000, x_food=10, y_food=20, Should_I_Plot=True):
    success_bool = 0             #becomes 1 when ant reaches the food
    s=1                          #distance moved by ant in one step
    x,y = 0,0                    #start location of our ant
    success_count = 0            #Used later for counting no. of successful ants
    no_of_steps = 0              #Counts no. of steps
    
    #Let's store our (x,y) data in list so that we can make a nice plot out of it.
    X=[x]
    Y=[y]

    #mu should depend upon theta. theta is inclinattion of d with x-axis. Define mu=theta_target
    #standard deviation is a function of distance b/w ant and food. sigma = m*d
    #theta = np.random.randn()    

    if np.sqrt((x_food-x)**2 + (y_food-y)**2)<1:
            print("ant's already standin' on the food!")

    for i in range(N):
        d = np.sqrt((x_food-x)**2 + (y_food-y)**2)
        sigma = m*d
        theta_target = np.arctan2(y_food-y,x_food-x)
        theta = np.random.normal(loc=theta_target, scale=sigma)
        dx = s*np.cos(theta)
        dy = s*np.sin(theta)
        x = x + dx
        y = y + dy
        X.append(x)
        Y.append(y)

        if d<1:
                success_bool = 1
                break
        no_of_steps+=1
        
    #Let us round our lists X and Y for the sake of gentle plotting
    X = np.round(X, 2).tolist()
    Y = np.round(Y, 2).tolist()

    #Finally creating a clean plot.
    if Should_I_Plot == True:
        plt.figure(figsize=(6, 6))
        plt.plot(X, Y, color='blue', linestyle='-', marker='.', label='Ant Path')
        plt.scatter(X[0], Y[0], color='green', s=100, label='Start')
        plt.scatter(X[-1], Y[-1], color='red', s=100, label='Food')
        plt.title('Ants path to food')
        plt.xlabel('X Position')
        plt.ylabel('Y Position')
        plt.grid(False)
        plt.legend()
        plt.show()
    
    return success_bool, no_of_steps, X, Y
````

````python
#Part 2
success_bool, no_of_steps, X_coord, Y_coord = ant_walker_function(m=0.1, N=1000, x_food=10, y_food=20, Should_I_Plot=True)
print("done in ", no_of_steps, " steps")
````

````python
#Part 3

#Let's assume that out of 1000 ants, atleast 100 of them will be successful (for reasionable convergable)

#Define a multi_run function that we can readily use in Part 6
def multi_run(x=10, y=20):
    count = 0
    no_of_runs = 0
    list_steps = [] #Define an empty list of no. of steps (We will fill this list in a moment)
    set_case=True   #This is true if we have 100 successful ants within first 1000 runs
    
    for i in range(1000):
        no_of_runs += 1
        success_bool, no_of_steps, X_coord, Y_coord=ant_walker_function(m=0.1,N=1000,x_food=x,y_food=y,Should_I_Plot=False)
        list_steps.append(no_of_steps)

        if success_bool == +1:
            count+=1
        if count == 100:
            break
        if i==999 and count<100:
            print("Bad Convergence. Better reduce m or distance.")
            set_case=False
    return no_of_runs, list_steps, count, set_case
        
#Now if we have a good convergent case, then we can calculate mean and standard deviation from list_steps

def mean_and_std_dev_extractor():
    if set_case==True:
        sum = 0
        squares_summed = 0
        for i in range(len(list_steps)):

            sum += list_steps[i]
            squares_summed += (list_steps[i])**2 

            mean = sum/len(list_steps)
            std_dev = np.sqrt(squares_summed/len(list_steps))
    else:
        print("Please select a reasionably small m and distance!")
        return None, None
    
    return mean, std_dev

no_of_runs, list_steps, count, set_case = multi_run()
mean, std_dev = mean_and_std_dev_extractor()
````

````python
#Part 4

#In Part 3, we defined a no_of_runs function which just calculates how many total times the for loop looped.
if set_case == True:
    print("No. of successful ants were" , count)
    print("This was from first ", no_of_runs, "ants who attempted the food chase challange!")
    print("So, the no. of unsuccessful ants are ", no_of_runs - count)
else:
    print("Please select a reasionably small m and distance!")
````

````python
#Part 5

#Now we plot no. of steps v/s how many ants did that kinda histogram.
#We take data directly from the list_steps defined in Part 3.
if set_case==True:
    plt.figure(figsize=(9,4))
    plt.hist(list_steps, bins=25, color="skyblue", edgecolor="k")
    plt.title('No. of Steps v/s How many Ants Histogram', fontsize=14)
    plt.xlabel('No. of Steps', fontsize=12)
    plt.ylabel('How many Ants', fontsize=12)
    plt.grid(False) # Optional horizontal gridlines
    plt.show()
else:
    print("Please select a reasionably small m and distance!")
````

````python
#Part 5

#d1[0,5] #d2[5,5] #d3[10,5] #d2[5,10] #d1[10,20] 
d1=np.sqrt(0*0 + 5*5)
d2=np.sqrt(5*5 + 5*5)
d3=np.sqrt(10*10 + 5*5)
d4=np.sqrt(5*5 + 10*10)
d5=np.sqrt(10*10 + 20*20)

distance_list=[d1,d2,d3,d4,d5]
mean_list=[]
std_dev_list=[]

no_of_runs, list_steps, count, set_case = multi_run(0,5)
mean, std_dev = mean_and_std_dev_extractor()
mean_list.append(mean)
std_dev_list.append(std_dev)

no_of_runs, list_steps, count, set_case = multi_run(5,5)
mean, std_dev = mean_and_std_dev_extractor()
mean_list.append(mean)
std_dev_list.append(std_dev)

no_of_runs, list_steps, count, set_case = multi_run(10,5)
mean, std_dev = mean_and_std_dev_extractor()
mean_list.append(mean)
std_dev_list.append(std_dev)

no_of_runs, list_steps, count, set_case = multi_run(5,10)
mean, std_dev = mean_and_std_dev_extractor()
mean_list.append(mean)
std_dev_list.append(std_dev)

no_of_runs, list_steps, count, set_case = multi_run(10,20)
mean, std_dev = mean_and_std_dev_extractor()
mean_list.append(mean)
std_dev_list.append(std_dev)

plt.errorbar(
    distance_list, 
    mean_list, 
    yerr=std_dev_list, 
    fmt='-o',          # Line style ('-') with point markers ('o')
    color='blue', 
    ecolor='red',      # Color of the error bars
    elinewidth=1.5,    # Thickness of the error bar lines
    capsize=5,         # Width of horizontal caps on top/bottom of error bars
    label='Ant Path (Mean ± Std Dev)'
)

plt.title('No. of steps v/s distance')
plt.xlabel('distance')
plt.ylabel('No. of steps')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()
````
